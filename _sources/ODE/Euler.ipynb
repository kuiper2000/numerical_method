```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for Euler method
dx = 0.01  # Step size
x_span = (0, 2)  # Solve from x = 0 to x = 2
n_steps = int((x_span[1] - x_span[0]) / dx)  # Number of steps

# Initial condition
y_0 = 1.0  # y(0)

# Arrays to store the results
y = np.zeros(n_steps)
x = np.linspace(x_span[0], x_span[1], n_steps)

# Assign initial condition
y[0] = y_0

# Euler method loop
for i in range(1, n_steps):
    y[i] = y[i-1] + dx * (np.exp(x[i-1]) - 4 * y[i-1])

# Plotting the result
plt.figure(figsize=(10, 6))
plt.plot(x, y, label="y(x) using Euler's Method")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Solution to ODE $y' + 4y = e^x$")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for Euler method
dt     = 0.01  # Time step
t_span = (0, 2)  # Solve from x = 0 to x = 2
n_steps = int((t_span[1] - t_span[0]) / dt)  # Number of steps

# Initial conditions
y1_0 = 0.0  # y(0)
y2_0 = 1.0  # y'(0)

# Arrays to store the results
y1 = np.zeros(n_steps)
y2 = np.zeros(n_steps)
x = np.linspace(t_span[0], t_span[1], n_steps)

# Assign initial conditions
y1[0] = y1_0
y2[0] = y2_0

# Euler method loop
for i in range(1, n_steps):
    y1[i] = y1[i-1] + dt * y2[i-1]
    y2[i] = y2[i-1] + dt * (np.exp(x[i-1]) - 4 * y2[i-1])

# Plotting the result
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label="y(x) using Euler's Method")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Solution to ODE $y'' + 4y' = e^x$")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def euler_method_ode2(y1_prime_func, y2_prime_func, y1_0, y2_0, t_span, dt):
    """
    Solve a second-order ODE using the Euler method.

    Parameters:
    - y1_prime_func: Function for y1' (y' = y2)
    - y2_prime_func: Function for y2' (y'' = f(x, y1, y2))
    - y1_0: Initial condition for y1
    - y2_0: Initial condition for y2
    - t_span: Tuple (t_start, t_end) for the range of x
    - dt: Time step for the Euler method

    Returns:
    - x: Array of x values
    - y1: Array of y1 values
    - y2: Array of y2 values
    """
    n_steps = int((t_span[1] - t_span[0]) / dt)  # Number of steps

    # Arrays to store the results
    y1 = np.zeros(n_steps)
    y2 = np.zeros(n_steps)
    x = np.linspace(t_span[0], t_span[1], n_steps)

    # Assign initial conditions
    y1[0] = y1_0
    y2[0] = y2_0

    # Euler method loop
    for i in range(1, n_steps):
        y1[i] = y1[i-1] + dt * y1_prime_func(x[i-1], y1[i-1], y2[i-1])
        y2[i] = y2[i-1] + dt * y2_prime_func(x[i-1], y1[i-1], y2[i-1])

    return x, y1, y2

# Example usage of the generalized Euler method

# Define the function for y1' (which is simply y2)
def y1_prime(x, y1, y2):
    return y2

# Define the function for y2' (the second derivative function)
def y2_prime(x, y1, y2):
    return np.exp(x) - 4 * y2  # This matches the original example

# Parameters
dt = 0.01
t_span = (0, 2)
y1_0 = 0.0  # y(0)
y2_0 = 1.0  # y'(0)

# Solve the ODE
x, y1, y2 = euler_method_ode2(y1_prime, y2_prime, y1_0, y2_0, t_span, dt)

# Plotting the result
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label="y(x) using Euler's Method")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Solution to ODE $y'' + 4y' = e^x$")
plt.legend()
plt.grid(True)
plt.show()
```
